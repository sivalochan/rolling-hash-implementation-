```python

class RollingHash:
    
    def __init__(self, s, base=131, mod=1000000007):
        
        self.s = s
        self.base = base
        self.mod = mod
        n = len(s)
        self.powers = [1] * (n + 1)
        self.prefix = [0] * (n + 1)
        for i in range(n):
            self.powers[i + 1] = (self.powers[i]*base)%mod
            value = ord(s[i])
            self.prefix[i + 1] = (self.prefix[i]*base + value)%mod

    def hash(self, left, right):
        length = right - left
        result = (self.prefix[right] - self.prefix[left]*self.powers[length])%self.mod
        return result

    def full_hash(self):
        return self.prefix[len(self.s)]
def rolling_hash(s, base=131, mod=1000000007):
    ans = 0
    for ch in s:
        ans = (ans * base + ord(ch)) % mod
    return ans



s = "abcd"

rh = RollingHash(s, base=131)

print("Full hash:", rh.full_hash())
print("Hash of 'bc':", rh.hash(1, 3))
print("Direct hash of 'bc':", rolling_hash("bc", base=131))
```
